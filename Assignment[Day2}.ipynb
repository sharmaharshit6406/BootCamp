```python
# 2d data using numpy
import numpy as np

# Creating 2-dimensional data
data = np.random.randint(1, 100, size=(5, 3))

print("2-Dimensional Data:")
print(data)
```

```python
#cleaning data using pandas
import pandas as pd

data = {"Name": ["John123", "mohan@45", "Jack#67", "David!!", "Ayush_89"],
    "Age": [20, 21, None, 22, 20]}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Remove unwanted characters from Name
df["Name"] = df["Name"].str.replace(r"[^A-Za-z]", "", regex=True)

# Fill missing age
df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\nCleaned Data:")
print(df)
```

```python
# Sine wave data using matplotlib
import numpy as np
import matplotlib.pyplot as plt

# Create x values
x = np.linspace(0, 2 * np.pi, 100)

# Calculate sine values
y = np.sin(x)

# Draw sine wave
plt.plot(x, y)

plt.xlabel("X")
plt.ylabel("sin(X)")
plt.title("Sine Wave")

plt.grid()
plt.show()
```
