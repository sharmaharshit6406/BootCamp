```python
##word cloud generation
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# Ensure necessary NLTK data is downloaded
nltk.download('punkt')
nltk.download('stopwords')
```

```python
text = """The environment is an essential part of our life because it provides us with air, water,
 food, and natural resources. It includes everything around us, such as plants, animals, rivers, forests,
and the atmosphere. Today, pollution, deforestation, climate change, and excessive use of natural
resources are causing serious environmental problems. We can protect the environment by planting trees, 
reducing plastic use, saving water and electricity, recycling waste, and using public transport when 
possible. Protecting the environment is the responsibility of every individual because a clean and 
healthy environment is necessary for a better future. """
print(text)
```

```python
words = word_tokenize(text.lower())  # Convert to lowercase for uniformity
print(words)
```

```python
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]
# The general syntax for a list comprehension is: [expression for item in iterable if condition]
print(filtered_words)
```

```python
word_freq = Counter(filtered_words) 
print(word_freq)
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)
```

```python
# Plotting the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Hide axes
plt.show()
```

```python
## Image editing using OpenCV

import cv2

# Read the image
img = cv2.imread("images.jpeg")

# Check if image is loaded
if img is None:
    print("Image not found!")
else:
    # 1. Resize image
    resized = cv2.resize(img, (500, 400))

    # 2. Crop image
    cropped = img[50:300, 50:400]

    # 3. Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 4. Rotate image
    rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    # 5. Add text
    text_img = img.copy()
    cv2.putText(text_img, "My Image", (50, 50),
                cv2.FONT_HERSHEY_SIMPLEX, 1,
                (0, 255, 0), 2)

    # 6. Draw a rectangle
    cv2.rectangle(text_img, (50, 80), (300, 250),
                  (255, 0, 0), 3)

    # Display images
    cv2.imshow("Original", img)
    cv2.imshow("Resized", resized)
    cv2.imshow("Cropped", cropped)
    cv2.imshow("Grayscale", gray)
    cv2.imshow("Rotated", rotated)
    cv2.imshow("Edited Image", text_img)

    # Wait for a key and close windows
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
