```python
##Creating  a Dataset
import numpy as np

players = np.array(['sachin','virat kohli','jaspreet','rishab','abhishek'])
matches = np.array(['IPL 2015', 'IPL 2020', 'IPL 2023'])

#np.random.seed(42)  # For reproducibility
runs = np.random.randint(10, 100, size=(len(players), len(matches)))

# Print the dataset
print("Players:", players)
print("Matches:", matches)
print("Runs scored by each player:")
print(runs)
```

```python
# Calculate total runs scored by each player
total_runs_by_player = runs.sum(axis=1)
print("Total runs scored by each player:")
for player, total in zip(players, total_runs_by_player):
    print(f"{player}: {total} runs")
```

```python
# Calculate average runs per match
average_runs_per_match = runs.mean(axis=0)
print("Average runs scored in each match:")
for match, avg in zip(matches, average_runs_per_match):
    print(f"{match}: {avg:.2f} runs")

# Find the player with the highest total runs
most_runs_index = np.argmax(total_runs_by_player)
print(f"The player with the most runs is {players[most_runs_index]} with {total_runs_by_player[most_runs_index]} runs.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate the engineering data
# 500 points from 0 to 10 seconds
time = np.linspace(0, 10, 500)

# Damped sine wave
amplitude = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)

# 2. Create the Canvas (Figure) and Coordinate Box (Axes) explicitly
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)

# 3. Plot the data with customized engineering styles
ax.plot(
    time,
    amplitude,
    color='crimson',
    linewidth=2,
    label='Sensor A Displacement'
)

# 4. Professional Engineering Customizations
ax.set_title(
    'Damped Harmonic Response of Structure',
    fontsize=14,
    fontweight='bold',
    pad=15
)

ax.set_xlabel(
    'Time (seconds)',
    fontsize=12
)

ax.set_ylabel(
    'Displacement (mm)',
    fontsize=12
)

# Enable engineering grid lines
ax.grid(
    True,
    linestyle='--',
    alpha=0.6
)

# Configure the Legend
ax.legend(
    loc='upper right',
    frameon=True,
    shadow=True
)

# 5. Improve layout
plt.tight_layout()

# 6. Display the graph
plt.show()
```

```python
## Harmonic Structure Response
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate engineering data using NumPy
time = np.linspace(0, 10, 500)  # 500 points from 0 to 10 seconds
amplitude1 = np.sin(2 * np.pi * time)  # Simple sine wave
amplitude2 = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)  # Damped sine wave

# 2. Create a figure with 2 subplots vertically (2 rows, 1 column)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), dpi=100)

# 3. Plot 1: Simple Sine Wave (Top Subplot)
ax1.plot(
    time,
    amplitude1,
    color="crimson",
    linewidth=2,
    label="Undamped Harmonic Motion",
)
ax1.set_title(
    "Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax1.set_xlabel("Time (seconds)", fontsize=11)
ax1.set_ylabel("Displacement (mm)", fontsize=11)
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend(loc="upper right", frameon=True, shadow=True)

# 4. Plot 2: Damped Sine Wave (Bottom Subplot)
ax2.plot(
    time, amplitude2, color="navy", linewidth=2, label="Damped Harmonic Motion"
)
ax2.set_title(
    "Damped Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax2.set_xlabel("Time (seconds)", fontsize=11)
ax2.set_ylabel("Displacement (mm)", fontsize=11)
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend(loc="upper right", frameon=True, shadow=True)

# 5. Adjust layout to avoid overlapping labels and display
plt.tight_layout()
# plt.savefig('sensor_data_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```
