```python
# program a)
#Random numbers and even numbers
import random

# Create a list of 10 random numbers
numbers = [random.randint(1, 100) for _ in range(10)]

print("Random numbers:", numbers)

# Print all even numbers
even_numbers = [num for num in numbers if num % 2 == 0]

print("Even numbers:", even_numbers)
```

```python
#program b)
# Reverse a string and reverse the words

# Create a string
text = "Python is a very powerful language"

# Reverse the complete string
reverse_string = text[::-1]

print("Original string:", text)
print("String in reverse order:", reverse_string)

# Reverse the order of words
reverse_words = " ".join(text.split()[::-1])

print("Words in reverse order:", reverse_words)
```
